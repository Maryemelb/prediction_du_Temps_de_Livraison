```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
fashion_mnist = keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()
```

```python
#construire_list
# Names of the integer classes, i.e., 0 -> T-short/top, 1 -> Trouser, etc.
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']
#display images with its name
plt.figure(figsize=(15,20))
for i in range(50):
    plt.subplot(10,5,i+1)
    plt.xticks([]) #remooves x numbers
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_images[i], cmap=plt.cm.binary)
    plt.xlabel(class_names[train_labels[i]])
plt.show()
```

```python
plt.figure(figsize=(15,20))
for i in range(50):
    plt.subplot(10,5,i+1)
    plt.xticks([]) #remooves x numbers
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_images[i], cmap=plt.cm.binary)
    plt.xlabel(train_labels[i])
plt.show()
```

Data preparation => already prepared

Normaliser les **pixels**

```python
train_images= train_images / 255.0
test_images = test_images /255.0
```

selects the first 10 elements of the array so we don't need d'utiliser les labels entiers (0–9)

```python
print(train_labels[:10])
#Keep the labels as integers (0–9), because you’ll use a loss function designed for integer labels =Q> labels already between 0-9
```

**Construction du modèle**

```python
model = tf.keras.Sequential([
    #transform data into single column
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    # Dense(neurals number , activation function)
    tf.keras.layers.Dense(128, activation='relu'),#128 : nbr neurons => whenever we add neurons its better /relu kathiyed lina linearité
    tf.keras.layers.Dense(10, activation ='softmax') #10 class to predict softmax returns probability
])
```

```python
model.summary()
```

Compiler

configure the model for training

```python
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

#adam: Adam is one of the most popular optimizers because : adjust weights efficiently =>know how it works
#Its job is to minimize the loss by changing the weights
#SparseCategoricalCrossentropy : used for multi-class classification and mesure the error
```

EarlyStopping to stop training when validation performance stops improving.



```python
from keras.callbacks import EarlyStopping
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    mode='min',
    restore_best_weights=True
)
history = model.fit(
    train_images, train_labels,
    epochs=10,
    validation_split=0.1,
    callbacks=[early_stop]
)
```

Train the model

```python

model.fit(train_images, train_labels, epochs=7,callbacks=[early_stop]) #epochs : how many time he will repeat the training
```

**Évaluation**

```python
labels_predict= model.predict(test_images)
print(labels_predict)
```

Evaluate accuracy

```python
#EVALUAT / calculate Accurancy and loss function
test_loss, test_acc = model.evaluate(test_images,  test_labels, verbose=2)
```
